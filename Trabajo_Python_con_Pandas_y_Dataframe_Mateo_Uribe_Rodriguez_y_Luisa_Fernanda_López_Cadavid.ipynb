```python
from ast import Index

# Definir las categorías de síntomas
df['Category'] = df['Sintoma'].apply(
    lambda x: 'Grave' if x in ['Infecciones bacterianas', 'Infección urinaria', 'Presión arterial alta',
                                'Infección respiratoria', 'Insuficiencia cardíaca', 'Hipertensión',
                                'Infección de piel', 'Depresión', 'Hiperglucemia']
    else ('Moderado' if x in ['Dolor severo', 'Asma', 'Inflamación',
                               'Dolor torácico', 'Retención de líquidos',
                               'Inflamación crónica', 'Bajo nivel de tiroides',
                               'Prevención de coágulos']
          else 'Leve')
)

# Mostrar valores por etiquetas
print(df.loc[df['Category'] == 'Grave'])
print(df.loc[df['Category'] == 'Moderado'])
print(df.loc[df['Category'] == 'Leve'])
# Mostrar valores por índice
print(df.iloc[7])
```

```python
df['Precio'] = [15.00, 12.00, 25.00, 40.00,
    15.00, 25.00, 20.00, 30.00,
    15.00, 15.00, 10.00, 20.00,
    25.00, 35.00, 20.00, 50.00,
    15.00, 10.00, 50.00, 20.00,
    15.00, 30.00, 30.00, 20.00,
    15.00]
df['Moneda'] = 'USD'
print(df)
```

```python
df['Status'] = np.random.choice(['Vivo', 'Muerto'], size=len(df))


df.dropna(inplace=True)
print(df)
```
